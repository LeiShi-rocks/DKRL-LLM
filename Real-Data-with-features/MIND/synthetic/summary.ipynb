```python
import numpy as np
import pandas as pd

# 1. Load and build `df` exactly as before
results = np.load("MC_simulation_results_fixed_penalty_full.npy",
                  allow_pickle=True).item()

rows = []
for q_decay, methods in results.items():
    for method, errs in methods.items():
        arr = np.array(errs)
        mse_train, mse_test = arr[:,0], arr[:,1]
        rmse_train = np.sqrt(mse_train)
        rmse_test  = np.sqrt(mse_test)
        rows.append({
            'Decay (q)': q_decay,
            'Method': method,
            'Train RMSE (mean)': rmse_train.mean(),
            'Train RMSE (std)':  rmse_train.std(),
            'Test RMSE (mean)':  rmse_test.mean(),
            'Test RMSE (std)':   rmse_test.std(),
        })

df = pd.DataFrame(rows)

# 2. Collect unique decays and methods
decays  = sorted(df['Decay (q)'].unique())
methods = df['Method'].unique()

# 3. Build train_summary with "mean (std)" strings
train_summary = pd.DataFrame(index=decays)
for method in methods:
    sub = df[df['Method']==method].set_index('Decay (q)')
    train_summary[method] = sub.apply(
        lambda row: f"{row['Train RMSE (mean)']:.3f} ({row['Train RMSE (std)']:.3f})",
        axis=1
    )

# 4. Build test_summary likewise
test_summary = pd.DataFrame(index=decays)
for method in methods:
    sub = df[df['Method']==method].set_index('Decay (q)')
    test_summary[method] = sub.apply(
        lambda row: f"{row['Test RMSE (mean)']:.3f} ({row['Test RMSE (std)']:.3f})",
        axis=1
    )

# 5. Clean up index name and display
train_summary.index.name = 'Decay (q)'
test_summary.index.name  = 'Decay (q)'

print("=== Train RMSE (mean ± std) ===")
print(train_summary)

print("\n=== Test RMSE (mean ± std) ===")
print(test_summary)

```
